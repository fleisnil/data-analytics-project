## Inferential tests
ANOVA compares mean log-delay across regions and reports eta-squared. Chi-squared tests the association between mode and a delay of at least five minutes and reports Cramér's V. Spearman correlation quantifies the monotonic precipitation-delay association. Large samples can make small effects statistically significant, so effect size and practical importance must be discussed.

```python
tests = pd.read_csv(PATHS.tables / 'statistical_tests.csv')
display(tests)
```
