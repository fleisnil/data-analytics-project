# 04 — Modeling, evaluation and additional analyses

The primary task is association analysis, not a standard prediction competition. OLS terms compare regions, modes and time periods while controlling for observed weather. A random forest is evaluated out of sample against overall and group-mean baselines fitted on training data only. K-means and a geographic map provide separate bonus analyses.

```python
import pandas as pd
from IPython.display import Image, HTML, display
from sptdelays.settings import PATHS
from sptdelays.modeling import run_models
from sptdelays.clustering import run_clustering
from sptdelays.geo import create_map
```

## Regression and evaluation
OLS uses `log1p(delay_minutes)` with station-clustered standard errors when possible; inspect the model specification and small-cluster warnings. Supported interactions compare regional mode associations. The random forest predicts minutes directly, with preprocessing fitted only on training data and a chronological holdout. One-day runs remain development checks.

```python
run_models()
display(pd.read_csv(PATHS.tables / 'model_metrics.csv'))
display(pd.read_csv(PATHS.tables / 'subgroup_metrics.csv').head(30))
display(pd.read_csv(PATHS.tables / 'holdout_coverage.csv'))
display(pd.read_csv(PATHS.tables / 'permutation_importance.csv').head(15))
```

```python
ols = pd.read_csv(PATHS.tables / 'ols_associations.csv')
display(ols)  # include confidence intervals and nonsignificant terms
print((PATHS.tables / 'model_specification.json').read_text(encoding='utf-8'))
```

Interpret categorical coefficients relative to their reference category. With a log1p outcome, exp(beta) describes a ratio of modeled geometric means of delay + 1, not an exact percentage change in delay minutes. Interactions require combined terms. Negative R² means worse squared error than the test-set mean; compare both training-only baselines and inspect their group fallback counts. This is retrospective association analysis with observed/grid weather, not a validated real-time forecast system.
