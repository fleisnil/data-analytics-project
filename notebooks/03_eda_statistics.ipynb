## Inferential tests
ANOVA compares mean delay in minutes across groups and reports eta-squared. Chi-squared tests group association with the five-minute delay indicator and reports Cramér's V. Spearman tests monotonic weather-delay associations. Inspect raw and Holm-adjusted p-values, expected cell counts and test warnings. These exploratory tests do not fully account for dependent calls; effect size and practical importance matter.

```python
tests = pd.read_csv(PATHS.tables / 'statistical_tests.csv')
display(tests)
display(pd.read_csv(PATHS.tables / 'regional_weighting_comparison.csv'))
display(pd.read_csv(PATHS.tables / 'weather_coverage_by_group.csv'))
display(pd.read_csv(PATHS.tables / 'daily_coverage_delay.csv'))
```
